```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive

drive.mount('/content/drive')

plt.style.use('ggplot')
```

```python
df = pd.read_csv('/content/drive/MyDrive/antropometria-dataset-1-1.csv')

df.head()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['peso'].dropna(), bins=15)
plt.title('Distribución del peso')
plt.xlabel('Peso (kg)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
import seaborn as sns

sns.scatterplot(data=df, x='peso', y='cintura', hue='sexo')
plt.show()
```

```python
hombre_cintura = df[df['sexo'] == 1]['cintura'].dropna()

plt.figure(figsize=(8, 4))
plt.boxplot(hombre_cintura, vert=False)
plt.title('Cintura de hombres')
plt.xlabel('Cintura (cm)')
plt.show()
```

```python
hombres = df[df['sexo'] == 1]

plt.figure(figsize=(8, 5))
plt.scatter(hombres['peso'], hombres['cintura'], s=5)
plt.title('Relación peso y cintura en hombres')
plt.xlabel('Peso (kg)')
plt.ylabel('Cintura (cm)')
plt.show()
```

```python
hombre_cintura = df[df['sexo'] == 1]['cintura'].dropna()

plt.figure(figsize=(8, 5))
plt.hist(hombre_cintura, bins=20)
plt.title('Distribución de cintura en hombres')
plt.xlabel('Cintura (cm)')
plt.ylabel('Frecuencia')
plt.show()
```
