```python
import pandas as pd

ruta = "/content/drive/MyDrive/antropometria-dataset-2.csv"
datos = pd.read_csv(ruta)

display(datos.head())
```

```python
tabla = pd.pivot_table(
    datos,
    index="desc_ent",
    values="peso",
    aggfunc="mean"
)

promedio_zacatecas = tabla.loc["ZACATECAS", "peso"]

print("Promedio exacto:", promedio_zacatecas)
print(f"Promedio con dos decimales: {promedio_zacatecas:.2f}")
```

```python
tabla_mediana = pd.pivot_table(
    datos,
    index="desc_ent",
    columns="sexo",
    values="peso",
    aggfunc="median"
)

display(tabla_mediana.loc[["YUCATAN"]])

print("Respuesta:", tabla_mediana.loc["YUCATAN", 1])
```

```python
tabla_promedios = pd.pivot_table(
    datos,
    index="desc_ent",
    columns="sexo",
    values="peso",
    aggfunc="mean"
)

print("Estado:", tabla_promedios[2].idxmax())
print("Promedio máximo:", tabla_promedios[2].max())
```

```python
adult = pd.read_csv("/content/drive/MyDrive/adult.csv")

adult["education"] = adult["education"].str.strip()
adult["marital-status"] = adult["marital-status"].str.strip()
```

```python
tabla_horas = pd.pivot_table(
    adult,
    index="education",
    columns="marital-status",
    values="hours-per-week",
    aggfunc="mean"
)

resultado = tabla_horas.loc["HS-grad", "Divorced"]

print("Promedio:", resultado)
```

```python
for columna in ["sex", "marital-status", "education"]:
    adult[columna] = adult[columna].str.strip()

tabla_edad = pd.pivot_table(
    adult,
    index=["education", "marital-status"],
    columns="sex",
    values="age",
    aggfunc="median"
)

resultado = tabla_edad.loc[
    ("Bachelors", "Married-AF-spouse"), "Male"
]

print("Mediana de edad:", resultado)
```
