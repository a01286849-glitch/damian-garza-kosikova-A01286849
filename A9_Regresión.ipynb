```python
from google.colab import drive
import pandas as pd

drive.mount('/content/drive')

ruta_archivo = '/content/drive/MyDrive/antropometria-dataset-1.csv'
df = pd.read_csv(ruta_archivo)
df.head()
```

```python
import pandas as pd
import statsmodels.api as sm


df_clean = df.dropna(subset=['peso', 'cadera']).copy()

X = sm.add_constant(df_clean['peso'])
y = df_clean['cadera']

modelo = sm.OLS(y, X).fit()

predicciones = modelo.predict(X)
residuos = y - predicciones

mediana_residuos = residuos.median()
print(f"Mediana de los residuos: {mediana_residuos:.4f}")
```

```python
df_clean = df.dropna(subset=['peso', 'cadera'])

correlacion = df_clean['peso'].corr(df_clean['cadera'])
print(f"Coeficiente de correlación: {correlacion:.4f}")
```

```python
cols = ['diastol', 'sistol', 'edad', 'peso', 'cintura']
df_clean = df.dropna(subset=cols).copy()

X = sm.add_constant(df_clean[['sistol', 'edad', 'peso', 'cintura']])
y = df_clean['diastol']
modelo = sm.OLS(y, X).fit()

print(modelo.summary())
```

```python
df_clean = df.dropna(subset=['diastol', 'sistol']).copy()

X = sm.add_constant(df_clean['sistol'])
y = df_clean['diastol']

modelo = sm.OLS(y, X).fit()

print(f"Valor de b0 (intercepto): {modelo.params['const']:.4f}")
```

```python
predicciones = modelo.predict(X)
residuos = y - predicciones

mediana_residuos = residuos.median()
print(f"Mediana de los residuos: {mediana_residuos:.4f}")
```
